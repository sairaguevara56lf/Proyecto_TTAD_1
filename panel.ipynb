```python
import json

# Abre y carga data.json con codificación UTF-8
with open("data.json", "r", encoding="utf-8") as archivo:
    datos = json.load(archivo)

ventas = datos["ventas"]  
print(f"Total de registros cargados: {len(ventas)}")
```

```python
ingresos_por_producto = {} 
for venta in ventas:
    producto = venta["producto"]
    ingreso = venta["cantidad"] * venta["precio_unitario"]


    if producto not in ingresos_por_producto:
        ingresos_por_producto[producto] = 0.0

    ingresos_por_producto[producto] += ingreso


producto_top = max(ingresos_por_producto, key=ingresos_por_producto.get)
ingreso_top = ingresos_por_producto[producto_top]

print("Ingresos por producto:", ingresos_por_producto)
print(f"Producto con más ingresos: {producto_top} ({ingreso_top:.2f})")
```

```python
vendedor_objetivo = "Ana" 


ventas_filtradas = [
    venta for venta in ventas if venta["vendedor"] == vendedor_objetivo
]

print(f"Ventas de {vendedor_objetivo}: {len(ventas_filtradas)}")
for venta in ventas_filtradas:
    importe = venta["cantidad"] * venta["precio_unitario"]
    print(f"  - ID {venta['id']} | {venta['mes']} | {venta['producto']} x{venta['cantidad']} = {importe:.2f}")
```

```python
importes_ventas = [
    venta["cantidad"] * venta["precio_unitario"] for venta in ventas
]

total_facturado_panel = sum(importes_ventas)
num_ventas = len(ventas)
ticket_medio = total_facturado_panel / num_ventas if num_ventas > 0 else 0

print(f"Facturación total: {total_facturado_panel:.2f}")
print(f"Número de ventas: {num_ventas}")
print(f"Ticket medio: {ticket_medio:.2f}")
```
